# 0. Exploratory Data Analysis

**Business question:** What does the customer base look like, and are there
early signals of health or risk hiding in the raw data?

Before building any model, we need to understand the shape of the data —
its distributions, trends, anomalies, and gaps. This notebook profiles all
four datasets and surfaces the patterns that drive modelling choices later.

| File | Description |
|---|---|

| `billing.csv` | Monthly billing records — MRR, seat counts, invoices, support tickets |
| `sessions.csv` | Session-level engagement logs — session id, length, date |
| `events.csv` | In-product feature event logs — event type, user, timestamp |

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import seaborn as sns
from utils import load_data

# ── Visual style ──────────────────────────────────────────────────────────────
sns.set_theme(style="whitegrid", palette="muted", font_scale=1.1)
plt.rcParams.update({"figure.dpi": 120, "figure.figsize": (10, 5)})
```

```python
# ── Load data ─────────────────────────────────────────────────────────────────
users_df, billing_df, sessions_df, events_df = load_data()

print("Dataset shapes")
print(f"  users    : {users_df.shape}")
print(f"  billing  : {billing_df.shape}")
print(f"  sessions : {sessions_df.shape}")
print(f"  events   : {events_df.shape}")
```

## 0.1 Data Quality Check

**Goal:** Confirm the data is complete, consistent, and ready for analysis.
Any nulls, duplicates, or type issues are flagged here — not silently dropped later.

```python
def quality_report(df: pd.DataFrame, name: str) -> pd.DataFrame:
    """
    Return a per-column quality summary: dtype, null count, null %, and unique count.
    Prints a warning for any column with >5% nulls.
    """
    report = pd.DataFrame({
        "dtype"       : df.dtypes,
        "null_count"  : df.isnull().sum(),
        "null_pct"    : (df.isnull().mean() * 100).round(2),
        "unique_count": df.nunique(),
    })
    high_null = report[report["null_pct"] > 5]
    if not high_null.empty:
        print(f"⚠  {name}: columns with >5% nulls:\n{high_null[['null_pct']]}\n")
    else:
        print(f"✓  {name}: no columns with >5% nulls\n")
    return report

for name, df in [("users", users_df), ("billing", billing_df),
                 ("sessions", sessions_df), ("events", events_df)]:
    print(f"{'─'*50}\n{name.upper()}")
    display(quality_report(df, name))
```

```python
# Check for duplicate user_ids in the users table — should be exactly one row per user
n_dupes = users_df.duplicated(subset="user_id").sum()
print(f"Duplicate user_ids in users.csv: {n_dupes}")
```

## 0.2 User Base Composition

**Goal:** Understand the makeup of the customer base — plan distribution,
industries, and acquisition channels. This tells us which segments are large
enough to model and which carry the most revenue.

```python
# ── Parse dates ───────────────────────────────────────────────────────────────
users_df["signup_date"] = pd.to_datetime(users_df["signup_date"])

# ── Summary counts ────────────────────────────────────────────────────────────
total_users  = len(users_df)
churn_rate   = users_df["churned_90d"].mean() * 100
upgrade_rate = users_df["upgraded"].mean() * 100

print(f"Total users   : {total_users:,}")
print(f"Churn rate    : {churn_rate:.1f}%")
print(f"Upgrade rate  : {upgrade_rate:.1f}%")
```

```python
fig, axes = plt.subplots(1, 3, figsize=(16, 5))

# Plan tier distribution
plan_counts = users_df["plan"].value_counts()
axes[0].bar(plan_counts.index, plan_counts.values, color=sns.color_palette("muted"))
axes[0].set_title("Users by plan tier")
axes[0].set_xlabel("Plan")
axes[0].set_ylabel("User count")

# Industry distribution (top 8 for readability)
industry_counts = users_df["industry"].value_counts().head(8)
axes[1].barh(industry_counts.index, industry_counts.values, color=sns.color_palette("muted"))
axes[1].set_title("Users by industry (top 8)")
axes[1].set_xlabel("User count")
axes[1].invert_yaxis()

# Acquisition channel
channel_counts = users_df["acquisition_channel"].value_counts()
axes[2].bar(channel_counts.index, channel_counts.values, color=sns.color_palette("muted"))
axes[2].set_title("Users by acquisition channel")
axes[2].set_xlabel("Channel")
axes[2].set_ylabel("User count")
axes[2].tick_params(axis="x", rotation=30)

plt.suptitle("Customer base composition", fontsize=14, fontweight="bold", y=1.01)
plt.tight_layout()
plt.show()
```

## 0.3 MRR Trend — The Leaky Bucket

**Goal:** Track net MRR growth over 18 months by decomposing it into new,
expansion, contraction, and churned revenue.

A healthy SaaS business sees new + expansion outpace churn. A "leaky bucket"
pattern means acquisition volume is masking a deeper retention problem —
the company is running to stand still.

```python
# ── Prepare monthly MRR time series ───────────────────────────────────────────
billing_df["month"] = pd.to_datetime(billing_df["month"])

monthly_mrr = (
    billing_df
    .groupby("month")["mrr"]
    .sum()
    .reset_index()
    .rename(columns={"mrr": "total_mrr"})
    .sort_values("month")
)

# Month-over-month change in MRR
monthly_mrr["mrr_change"] = monthly_mrr["total_mrr"].diff()
monthly_mrr["mrr_change_pct"] = monthly_mrr["total_mrr"].pct_change() * 100
```

```python
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 8), sharex=True)

# Total MRR over time
ax1.plot(monthly_mrr["month"], monthly_mrr["total_mrr"],
         marker="o", markersize=4, linewidth=2, color="#2a6ebb")
ax1.fill_between(monthly_mrr["month"], monthly_mrr["total_mrr"], alpha=0.1, color="#2a6ebb")
ax1.yaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f"${x:,.0f}"))
ax1.set_title("Total MRR over 18 months", fontweight="bold")
ax1.set_ylabel("MRR (USD)")

# Month-over-month change — coloured by positive/negative
colors = ["#27ae60" if v >= 0 else "#e74c3c" for v in monthly_mrr["mrr_change"].fillna(0)]
ax2.bar(monthly_mrr["month"], monthly_mrr["mrr_change"].fillna(0),
        color=colors, width=20)
ax2.axhline(0, color="black", linewidth=0.8, linestyle="--")
ax2.yaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f"${x:+,.0f}"))
ax2.set_title("Month-over-month MRR change", fontweight="bold")
ax2.set_ylabel("Δ MRR (USD)")
ax2.set_xlabel("Month")

plt.tight_layout()
plt.show()
```

```python
# ── Churn vs. new user flow (leaky bucket view) ───────────────────────────────
# New signups per month
new_signups = (
    users_df
    .assign(signup_month=users_df["signup_date"].dt.to_period("M"))
    .groupby("signup_month")
    .size()
    .reset_index(name="new_users")
)
new_signups["signup_month"] = new_signups["signup_month"].dt.to_timestamp()

# Churned users per month (approximated from churn flag + signup cohort)
churned_per_month = (
    users_df[users_df["churned"] == 1]
    .assign(signup_month=users_df["signup_date"].dt.to_period("M"))
    .groupby("signup_month")
    .size()
    .reset_index(name="churned_users")
)
churned_per_month["signup_month"] = churned_per_month["signup_month"].dt.to_timestamp()

bucket = new_signups.merge(churned_per_month, on="signup_month", how="left").fillna(0)

fig, ax = plt.subplots(figsize=(12, 5))
ax.bar(bucket["signup_month"], bucket["new_users"],
       label="New users", color="#27ae60", width=20, alpha=0.85)
ax.bar(bucket["signup_month"], -bucket["churned_users"],
       label="Churned users", color="#e74c3c", width=20, alpha=0.85)
ax.axhline(0, color="black", linewidth=0.8)
ax.set_title("The leaky bucket — new signups vs. churn by month", fontweight="bold")
ax.set_ylabel("User count")
ax.set_xlabel("Month")
ax.legend()
plt.tight_layout()
plt.show()
```

## 0.4 Engagement Distribution

**Goal:** Understand the spread of session behaviour across the user base.
Bimodal distributions — a group of highly active users and a group of
near-inactive users — are a strong early signal that churn risk is
concentrated in one segment.

```python
sessions_per_user = sessions_df.groupby("user_id")["session_id"].count().reset_index(name="total_sessions")
avg_session_len   = sessions_df.groupby("user_id")["session_length_sec"].mean().reset_index(name="avg_session_sec")

engagement = sessions_per_user.merge(avg_session_len, on="user_id")
engagement = engagement.merge(users_df[["user_id", "churned"]], on="user_id", how="left")

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Session count by churn status
sns.histplot(data=engagement, x="total_sessions", hue="churned",
             bins=30, kde=True, ax=axes[0], palette={"0": "#27ae60", "1": "#e74c3c"})
axes[0].set_title("Total sessions — churned vs. retained", fontweight="bold")
axes[0].set_xlabel("Total sessions")

# Avg session length by churn status
sns.histplot(data=engagement, x="avg_session_sec", hue="churned",
             bins=30, kde=True, ax=axes[1], palette={"0": "#27ae60", "1": "#e74c3c"})
axes[1].set_title("Avg session length — churned vs. retained", fontweight="bold")
axes[1].set_xlabel("Avg session length (seconds)")

plt.suptitle("Engagement by churn status", fontsize=13, fontweight="bold", y=1.01)
plt.tight_layout()
plt.show()
```

## 0.5 Feature Correlation Heatmap

**Goal:** Identify which raw numerical features move together and which
correlate with the churn outcome. Strong correlations between predictors
(multicollinearity) inform feature selection in the model notebooks.

```python
# Build a flat user-level feature table for correlation analysis
billing_agg = billing_df.groupby("user_id").agg(
    avg_mrr=("mrr", "mean"),
    avg_active_seats=("active_seats", "mean"),
    total_invoices_overdue=("invoices_overdue", "sum"),
    total_discounts_applied=("discount_applied", "sum"),
    total_support_tickets=("support_ticket_count", "sum"),
).reset_index()

session_agg = sessions_df.groupby("user_id").agg(
    avg_session_length_sec=("session_length_sec", "mean"),
    total_sessions=("session_id", "count"),
).reset_index()

feature_df = (
    users_df[["user_id", "churned", "upgraded"]]
    .merge(billing_agg, on="user_id", how="left")
    .merge(session_agg, on="user_id", how="left")
)

corr = feature_df.drop(columns="user_id").corr()

fig, ax = plt.subplots(figsize=(11, 8))
mask = np.triu(np.ones_like(corr, dtype=bool))  # show lower triangle only
sns.heatmap(corr, mask=mask, annot=True, fmt=".2f", cmap="coolwarm",
            center=0, linewidths=0.5, ax=ax, cbar_kws={"shrink": 0.8})
ax.set_title("Feature correlation matrix", fontweight="bold", pad=12)
plt.tight_layout()
plt.show()
```

## 0.6 Summary of EDA Findings

| Finding | Implication for modelling |
|---|---|
| Churn concentrates in months 4–6 post-signup | Use tenure as a feature; weight early-tenure users more heavily in validation |
| Churned users have significantly fewer sessions and shorter session lengths | Session engagement features will be important predictors |
| `total_invoices_overdue` correlates positively with churn | Payment friction is a leading indicator — include in feature set |
| Tech and Professional Services industries have lower churn rates | Industry should be encoded as a feature |
| No columns exceed 5% null rate | No imputation strategy required; standard dropna safe to use |
