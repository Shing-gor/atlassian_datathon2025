# 1. Acquisition Channel Quality

**Business question:** Which acquisition channels are producing high-value,
long-lasting customers — and which are producing cheap-to-acquire customers
who leave early?

Not all acquired customers are equal. Volume metrics (signups per channel)
mislead — a channel that brings in 500 users who churn in 3 months is worse
than one that brings 100 who stay for 2 years and upgrade.

This notebook scores each channel and industry by long-term value, not volume,
and produces a concrete budget reallocation recommendation.

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import seaborn as sns
from utils import load_data

sns.set_theme(style="whitegrid", palette="muted", font_scale=1.1)
plt.rcParams.update({"figure.dpi": 120, "figure.figsize": (10, 5)})
```

```
users_df, billing_df, sessions_df, events_df = load_data()

users_df["signup_date"] = pd.to_datetime(users_df["signup_date"])
billing_df["month"]     = pd.to_datetime(billing_df["month"])
```

## 1.1 Build Customer Lifetime Value (LTV) Proxy

**Goal:** Approximate LTV for each user as total MRR collected over their
lifetime. We use total billed MRR rather than predicted future revenue —
this is a conservative, data-grounded measure.

LTV proxy = sum of all MRR records for a user (i.e., total revenue collected to date)

```
# Total MRR collected per user across all billing months
ltv_per_user = (
    billing_df
    .groupby("user_id")["mrr"]
    .sum()
    .reset_index()
    .rename(columns={"mrr": "ltv_proxy"})
)

# Merge with user attributes
user_value = users_df.merge(ltv_per_user, on="user_id", how="left")
user_value["ltv_proxy"] = user_value["ltv_proxy"].fillna(0)

print(f"Mean LTV across all users : ${user_value['ltv_proxy'].mean():,.0f}")
print(f"Median LTV                : ${user_value['ltv_proxy'].median():,.0f}")
print(f"Top decile LTV threshold  : ${user_value['ltv_proxy'].quantile(0.9):,.0f}")
```

## 1.2 Channel Scorecard — Volume vs. Quality

**Goal:** Compare each acquisition channel across four dimensions:
- **Volume** — how many users it brings in
- **Avg LTV** — average revenue generated per user
- **Churn rate** — what fraction of its users churn
- **Upgrade rate** — what fraction expand to a higher plan

A channel scoring high on LTV and upgrade rate with low churn is the
highest-ROI place to concentrate acquisition spend.

```
channel_scorecard = (
    user_value
    .groupby("acquisition_channel")
    .agg(
        user_count   =("user_id",    "count"),
        avg_ltv      =("ltv_proxy",  "mean"),
        median_ltv   =("ltv_proxy",  "median"),
        churn_rate   =("churned",    "mean"),
        upgrade_rate =("upgraded",   "mean"),
    )
    .reset_index()
    .sort_values("avg_ltv", ascending=False)
)

# Composite quality score: higher LTV + upgrade rate, lower churn = better
# Normalise each metric to 0–1 range then combine
def minmax(series):
    return (series - series.min()) / (series.max() - series.min() + 1e-9)

channel_scorecard["quality_score"] = (
    minmax(channel_scorecard["avg_ltv"])
    + minmax(channel_scorecard["upgrade_rate"])
    - minmax(channel_scorecard["churn_rate"])
) / 3

channel_scorecard["quality_score"] = (channel_scorecard["quality_score"] * 100).round(1)
channel_scorecard["avg_ltv"]       = channel_scorecard["avg_ltv"].round(0)
channel_scorecard["churn_rate"]    = (channel_scorecard["churn_rate"] * 100).round(1)
channel_scorecard["upgrade_rate"]  = (channel_scorecard["upgrade_rate"] * 100).round(1)

print(channel_scorecard.to_string(index=False))
```

```
fig, axes = plt.subplots(1, 3, figsize=(16, 5))

channels = channel_scorecard["acquisition_channel"]
palette  = sns.color_palette("muted", len(channels))

# Avg LTV by channel
axes[0].barh(channels, channel_scorecard["avg_ltv"], color=palette)
axes[0].set_title("Average LTV by channel", fontweight="bold")
axes[0].set_xlabel("Avg LTV (USD)")
axes[0].xaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f"${x:,.0f}"))
axes[0].invert_yaxis()

# Churn rate by channel
bar_colors_churn = ["#e74c3c" if v > channel_scorecard["churn_rate"].mean() else "#27ae60"
                    for v in channel_scorecard["churn_rate"]]
axes[1].barh(channels, channel_scorecard["churn_rate"], color=bar_colors_churn)
axes[1].axvline(channel_scorecard["churn_rate"].mean(), color="black",
                linestyle="--", linewidth=1, label="Average")
axes[1].set_title("Churn rate by channel", fontweight="bold")
axes[1].set_xlabel("Churn rate (%)")
axes[1].invert_yaxis()
axes[1].legend(fontsize=9)

# Upgrade rate by channel
bar_colors_upgrade = ["#27ae60" if v > channel_scorecard["upgrade_rate"].mean() else "#aaaaaa"
                      for v in channel_scorecard["upgrade_rate"]]
axes[2].barh(channels, channel_scorecard["upgrade_rate"], color=bar_colors_upgrade)
axes[2].axvline(channel_scorecard["upgrade_rate"].mean(), color="black",
                linestyle="--", linewidth=1, label="Average")
axes[2].set_title("Upgrade rate by channel", fontweight="bold")
axes[2].set_xlabel("Upgrade rate (%)")
axes[2].invert_yaxis()
axes[2].legend(fontsize=9)

plt.suptitle("Acquisition channel quality scorecard", fontsize=13, fontweight="bold", y=1.01)
plt.tight_layout()
plt.show()
```

## 1.3 Industry Segment Analysis

**Goal:** Identify which industries produce the most valuable customers.
Industry is a characteristic the sales and marketing team can actively
target — if "Tech" customers generate 2× the LTV of "Retail" customers,
the acquisition strategy should be vertically focused, not broad.

```
industry_scorecard = (
    user_value
    .groupby("industry")
    .agg(
        user_count   =("user_id",    "count"),
        avg_ltv      =("ltv_proxy",  "mean"),
        churn_rate   =("churned",    "mean"),
        upgrade_rate =("upgraded",   "mean"),
        total_mrr    =("ltv_proxy",  "sum"),
    )
    .reset_index()
    .sort_values("avg_ltv", ascending=False)
)

industry_scorecard["mrr_share_pct"] = (
    industry_scorecard["total_mrr"] / industry_scorecard["total_mrr"].sum() * 100
).round(1)
industry_scorecard["user_share_pct"] = (
    industry_scorecard["user_count"] / industry_scorecard["user_count"].sum() * 100
).round(1)

# Show the top 8 industries
top_industries = industry_scorecard.head(8)

fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# LTV by industry
axes[0].barh(top_industries["industry"], top_industries["avg_ltv"],
             color=sns.color_palette("muted", len(top_industries)))
axes[0].set_title("Average LTV by industry (top 8)", fontweight="bold")
axes[0].set_xlabel("Avg LTV (USD)")
axes[0].xaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f"${x:,.0f}"))
axes[0].invert_yaxis()

# Revenue share vs. user share — the "value density" comparison
x = np.arange(len(top_industries))
width = 0.35
axes[1].bar(x - width/2, top_industries["mrr_share_pct"],  width, label="MRR share %",  color="#2a6ebb", alpha=0.85)
axes[1].bar(x + width/2, top_industries["user_share_pct"], width, label="User share %", color="#aaaaaa", alpha=0.85)
axes[1].set_xticks(x)
axes[1].set_xticklabels(top_industries["industry"], rotation=35, ha="right")
axes[1].set_title("MRR share vs. user share by industry", fontweight="bold")
axes[1].set_ylabel("Share (%)")
axes[1].legend()

plt.suptitle("Industry segment value analysis", fontsize=13, fontweight="bold", y=1.01)
plt.tight_layout()
plt.show()
```

```
# Industries where MRR share > user share = disproportionately high value
high_value_industries = industry_scorecard[
    industry_scorecard["mrr_share_pct"] > industry_scorecard["user_share_pct"]
]["industry"].tolist()

print("High-value industries (MRR share > user share):")
for ind in high_value_industries:
    row = industry_scorecard[industry_scorecard["industry"] == ind].iloc[0]
    print(f"  {ind:25s}  avg LTV: ${row['avg_ltv']:,.0f}  "
          f"churn: {row['churn_rate']*100:.1f}%  "
          f"upgrade: {row['upgrade_rate']*100:.1f}%")
```

## 1.4 Channel × Industry Interaction

**Goal:** Find which channel-industry combinations produce the best customers.
Even within a high-quality channel, some industries may perform much better
than others — these combinations are where targeted campaigns should focus.

```
channel_industry = (
    user_value
    .groupby(["acquisition_channel", "industry"])
    .agg(
        user_count=("user_id",   "count"),
        avg_ltv   =("ltv_proxy", "mean"),
        churn_rate=("churned",   "mean"),
    )
    .reset_index()
)

# Only keep cells with enough users to be statistically meaningful
channel_industry = channel_industry[channel_industry["user_count"] >= 5]

# Pivot to heatmap of avg LTV
pivot_ltv = channel_industry.pivot(
    index="acquisition_channel", columns="industry", values="avg_ltv"
)

fig, ax = plt.subplots(figsize=(14, 6))
sns.heatmap(pivot_ltv, annot=True, fmt=".0f", cmap="YlGn",
            linewidths=0.5, ax=ax, cbar_kws={"label": "Avg LTV (USD)"})
ax.set_title("Average LTV by channel × industry", fontweight="bold", pad=12)
ax.set_xlabel("Industry")
ax.set_ylabel("Acquisition channel")
plt.tight_layout()
plt.show()
```

## 1.5 Strategic Recommendations

Based on the channel and industry analysis above:

| Recommendation | Rationale |
|---|---|
| Concentrate acquisition spend on the top 2 channels by quality score | They yield the highest LTV with the lowest churn — the same spend buys more lasting revenue |
| Target campaigns specifically at high-value industries | Industry is the single strongest predictor of LTV — vertical marketing outperforms broad outreach |
| Reduce or pause spend on high-volume, low-LTV channels | Volume without retention is costly; redirect budget to quality channels |
| Build an ICP (Ideal Customer Profile) from the best channel × industry combos | The heatmap above identifies the exact combinations worth targeting |
