# 3. Upgrade Propensity Model

**Business question:** Which customers on lower-tier plans are ready to
upgrade — and what does "ready" actually look like in the data?

Expansion revenue (upgrading existing customers) has near-zero acquisition
cost and significantly higher close rates than net-new business. Yet most
SaaS companies under-invest in this motion — the growth lever is sitting
untouched inside the existing base.

This notebook identifies customers who are hitting plan limits, showing
high engagement, and match the profile of users who have historically
upgraded. The output is a scored, prioritised list for the sales team.

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score
from sklearn.metrics import (
    classification_report, confusion_matrix, ConfusionMatrixDisplay,
    roc_auc_score, roc_curve
)
from sklearn.preprocessing import LabelEncoder
from utils import load_data, aggregate_billing, aggregate_sessions

sns.set_theme(style="whitegrid", palette="muted", font_scale=1.1)
plt.rcParams.update({"figure.dpi": 120, "figure.figsize": (10, 5)})

RANDOM_STATE = 42
```

```
users_df, billing_df, sessions_df, events_df = load_data()
```

## 3.1 Understanding Historical Upgraders

**Goal:** Before building a model, study the customers who *did* upgrade.
What did their behaviour look like in the 60 days before they upgraded?
This tells us what signals to include in the feature set and validates
that upgrading has a detectable behavioural fingerprint.

```
# Split into upgraders and non-upgraders
upgraders     = users_df[users_df["upgraded"] == 1]
non_upgraders = users_df[users_df["upgraded"] == 0]

upgrade_rate_overall = users_df["upgraded"].mean() * 100
print(f"Overall upgrade rate : {upgrade_rate_overall:.1f}%")
print(f"Total upgraders      : {len(upgraders):,}")
print(f"Total non-upgraders  : {len(non_upgraders):,}")
```

```
# Compare billing characteristics of upgraders vs. non-upgraders
billing_agg = aggregate_billing(billing_df)

billing_with_flag = billing_agg.merge(
    users_df[["user_id", "upgraded"]], on="user_id", how="left"
)

comparison_cols = [
    "avg_mrr", "avg_active_seats",
    "total_invoices_overdue", "total_support_tickets"
]

comparison = (
    billing_with_flag
    .groupby("upgraded")[comparison_cols]
    .mean()
    .T
    .rename(columns={0: "Non-upgrader avg", 1: "Upgrader avg"})
)
comparison["ratio (upgrader / non)"] = (
    comparison["Upgrader avg"] / comparison["Non-upgrader avg"]
).round(2)

print("\nBilling profile comparison — upgraders vs. non-upgraders:")
print(comparison.round(2).to_string())
```

```
# Visualise the differences
fig, axes = plt.subplots(2, 2, figsize=(13, 9))

for ax, col in zip(axes.flatten(), comparison_cols):
    plot_data = billing_with_flag[["upgraded", col]].copy()
    plot_data["upgraded"] = plot_data["upgraded"].map({0: "Non-upgrader", 1: "Upgrader"})

    sns.boxplot(data=plot_data, x="upgraded", y=col, ax=ax,
                palette={"Non-upgrader": "#aaaaaa", "Upgrader": "#27ae60"})
    ax.set_title(col.replace("_", " ").title(), fontweight="bold")
    ax.set_xlabel("")

plt.suptitle("Billing behaviour — upgraders vs. non-upgraders",
             fontsize=13, fontweight="bold", y=1.01)
plt.tight_layout()
plt.show()
```

## 3.2 Feature Engineering

The upgrade model uses the same feature pipeline as the churn model,
with one key addition: **seat utilisation rate** — customers approaching
their seat limit are prime upgrade candidates.

| Feature | Upgrade signal |
|---|---|
| `avg_active_seats` | High seat usage = approaching plan limits |
| `total_sessions` | High engagement = high perceived value |
| `unique_event_types` | Feature breadth = power users are more upgrade-ready |
| `avg_mrr` | Already paying well — expansion is a natural next step |
| `plan_encoded` | Only lower-tier users can upgrade |

```
session_agg = aggregate_sessions(sessions_df)

event_agg = (
    events_df
    .groupby("user_id")
    .agg(
        total_events       =("event_type", "count"),
        unique_event_types =("event_type", "nunique"),
    )
    .reset_index()
)

le_plan     = LabelEncoder()
le_industry = LabelEncoder()
le_channel  = LabelEncoder()

users_encoded = users_df.copy()
users_encoded["plan_encoded"]     = le_plan.fit_transform(users_df["plan"])
users_encoded["industry_encoded"] = le_industry.fit_transform(users_df["industry"])
users_encoded["channel_encoded"]  = le_channel.fit_transform(users_df["acquisition_channel"])

feature_df = (
    users_encoded[[
        "user_id", "upgraded",
        "plan_encoded", "industry_encoded", "channel_encoded"
    ]]
    .merge(billing_agg, on="user_id", how="left")
    .merge(session_agg, on="user_id", how="left")
    .merge(event_agg,   on="user_id", how="left")
    .fillna(0)
)

# Exclude users who have already churned — they cannot upgrade
feature_df = feature_df.merge(
    users_df[["user_id", "churned"]], on="user_id", how="left"
)
feature_df = feature_df[feature_df["churned"] == 0].drop(columns="churned")

print(f"Feature table shape   : {feature_df.shape}")
print(f"Upgrade rate in sample: {feature_df['upgraded'].mean()*100:.1f}%")
```

## 3.3 Model Training

Same Random Forest architecture as the churn model, again with
`class_weight="balanced"` because upgraders are a minority class.

```
FEATURE_COLS = [c for c in feature_df.columns if c not in ("user_id", "upgraded")]
TARGET_COL   = "upgraded"

X = feature_df[FEATURE_COLS]
y = feature_df[TARGET_COL]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=RANDOM_STATE, stratify=y
)

upgrade_model = RandomForestClassifier(
    n_estimators=300,
    max_depth=8,
    min_samples_leaf=10,
    class_weight="balanced",
    random_state=RANDOM_STATE,
    n_jobs=-1,
)
upgrade_model.fit(X_train, y_train)

# Cross-validated AUC
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
cv_auc = cross_val_score(upgrade_model, X_train, y_train, cv=cv, scoring="roc_auc")
print(f"Cross-validated AUC (5-fold): {cv_auc.mean():.3f} ± {cv_auc.std():.3f}")
```

## 3.4 Model Evaluation

```
y_pred      = upgrade_model.predict(X_test)
y_pred_prob = upgrade_model.predict_proba(X_test)[:, 1]

print("=" * 55)
print("UPGRADE MODEL — TEST SET PERFORMANCE")
print("=" * 55)
print(classification_report(y_test, y_pred, target_names=["Did not upgrade", "Upgraded"]))
print(f"AUC-ROC: {roc_auc_score(y_test, y_pred_prob):.3f}")
```

```
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

# Confusion matrix
cm = confusion_matrix(y_test, y_pred)
ConfusionMatrixDisplay(cm, display_labels=["Not upgraded", "Upgraded"]).plot(
    ax=axes[0], colorbar=False, cmap="Greens"
)
axes[0].set_title("Confusion matrix", fontweight="bold")

# ROC curve
fpr, tpr, _ = roc_curve(y_test, y_pred_prob)
axes[1].plot(fpr, tpr, linewidth=2, color="#27ae60",
             label=f"AUC = {roc_auc_score(y_test, y_pred_prob):.3f}")
axes[1].plot([0, 1], [0, 1], "k--", linewidth=1, label="Random")
axes[1].set_xlabel("False positive rate")
axes[1].set_ylabel("True positive rate")
axes[1].set_title("ROC curve — upgrade model", fontweight="bold")
axes[1].legend()

plt.tight_layout()
plt.show()
```

## 3.5 Feature Importance — What Signals Predict Upgrading?

```
importance_df = (
    pd.DataFrame({
        "feature"   : FEATURE_COLS,
        "importance": upgrade_model.feature_importances_,
    })
    .sort_values("importance", ascending=False)
    .reset_index(drop=True)
)

fig, ax = plt.subplots(figsize=(10, 6))
ax.barh(
    importance_df["feature"][::-1],
    importance_df["importance"][::-1],
    color=sns.color_palette("Greens_d", len(importance_df))
)
ax.set_title("Feature importance — upgrade model", fontweight="bold")
ax.set_xlabel("Mean decrease in impurity (normalised)")
plt.tight_layout()
plt.show()

print("\nTop 5 features driving upgrade propensity:")
print(importance_df.head(5).to_string(index=False))
```

## 3.6 Expansion Revenue Opportunity — Scoring the Current Base

**Goal:** Score every non-upgraded, non-churned user with their upgrade
probability. Customers who score above 0.6 AND are on a lower-tier plan
are the immediate expansion revenue opportunity.

```
# Score all eligible users (non-churned, not yet upgraded)
eligible = feature_df[feature_df["upgraded"] == 0].copy()
eligible["upgrade_probability"] = upgrade_model.predict_proba(
    eligible[FEATURE_COLS]
)[:, 1]

# Classify into propensity tiers
eligible["propensity_tier"] = pd.cut(
    eligible["upgrade_probability"],
    bins=[0, 0.3, 0.6, 1.0],
    labels=["Low", "Medium", "High"],
)

propensity_summary = eligible.groupby("propensity_tier", observed=True).agg(
    user_count        =("user_id", "count"),
    avg_upgrade_prob  =("upgrade_probability", "mean"),
).reset_index()

print("Upgrade propensity distribution across eligible base:")
print(propensity_summary.to_string(index=False))
```

```
# High-propensity users = immediate expansion opportunity
high_propensity = (
    eligible[eligible["propensity_tier"] == "High"]
    [["user_id", "upgrade_probability", "propensity_tier"]]
    .sort_values("upgrade_probability", ascending=False)
    .merge(
        users_df[["user_id", "plan", "industry", "acquisition_channel"]],
        on="user_id", how="left"
    )
    .reset_index(drop=True)
)

pct_of_base = len(high_propensity) / len(users_df) * 100
print(f"\nHigh-propensity upgrade candidates : {len(high_propensity):,} users")
print(f"As % of total user base            : {pct_of_base:.1f}%")
print(f"\nTop 20 priority accounts for expansion outreach:")
print(high_propensity.head(20).to_string(index=False))
```

```
# Estimate the revenue impact of converting high-propensity users
avg_mrr_after_upgrade = billing_df.groupby("user_id")["mrr"].mean().mean()
estimated_new_mrr = len(high_propensity) * avg_mrr_after_upgrade * 0.3  # conservative 30% uplift assumption

print(f"\nConservative expansion revenue estimate")
print(f"  High-propensity users               : {len(high_propensity):,}")
print(f"  Assumed avg MRR uplift on upgrade    : 30% of avg MRR (${avg_mrr_after_upgrade * 0.3:,.0f}/user/month)")
print(f"  Estimated monthly revenue opportunity: ${estimated_new_mrr:,.0f}")
print(f"  Estimated annual revenue opportunity : ${estimated_new_mrr * 12:,.0f}")
```

```
# Save the scored list
high_propensity.to_csv("upgrade_priority_list.csv", index=False)
print("\nSaved: upgrade_priority_list.csv")
```
